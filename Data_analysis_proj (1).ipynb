```python
print("Weather:", weather_clean.duplicated().sum())

print("Air Quality:", air_quality_clean.duplicated().sum())

print("Cities:", cities_clean.duplicated().sum())
```

```python
weather_clean = weather_clean.drop_duplicates()

air_quality_clean = air_quality_clean.drop_duplicates()

cities_clean = cities_clean.drop_duplicates()
```

```python
print(weather_clean.duplicated().sum())
print(air_quality_clean.duplicated().sum())
print(cities_clean.duplicated().sum())
```

```python
weather_clean['date'] = pd.to_datetime(
    weather_clean['date']
)

air_quality_clean['date'] = pd.to_datetime(
    air_quality_clean['date']
)
```

```python
print(weather_clean['date'].dtype)

print(air_quality_clean['date'].dtype)
```

```python
weather_clean['year'] = weather_clean['date'].dt.year

weather_clean['month'] = weather_clean['date'].dt.month

weather_clean['day'] = weather_clean['date'].dt.day
```

```python
print(
    "Weather duplicate city-date records:",
    weather_clean.duplicated(
        subset=['city', 'date']
    ).sum()
)

print(
    "Air Quality duplicate city-date records:",
    air_quality_clean.duplicated(
        subset=['city', 'date']
    ).sum()
)
```

```python
print("Weather date range:")
print(weather_clean['date'].min(), "to",
      weather_clean['date'].max())

print("\nAir Quality date range:")
print(air_quality_clean['date'].min(), "to",
      air_quality_clean['date'].max())
```

```python
weather_air = pd.merge(
    weather_clean,
    air_quality_clean,
    on=['city', 'date'],
    how='inner',
    validate='one_to_one'
)

display(weather_air.head())
```

```python
final_df = pd.merge(
    weather_air,
    cities_clean,
    on='city',
    how='left',
    validate='many_to_one'
)

display(final_df.head())
```

```python
city_summary = (
    final_df.groupby('city')
    .agg(
        Average_PM25=('pm2_5_mean', 'mean'),
        Total_Records=('pm2_5_mean', 'count')
    )
    .reset_index()
)

# Consider cities with at least 30 observations
city_summary = city_summary[
    city_summary['Total_Records'] >= 30
]

top_10_polluted = city_summary.sort_values(
    by='Average_PM25',
    ascending=False
).head(10)

display(top_10_polluted)
```

```python
bottom_10_polluted = city_summary.sort_values(
    by='Average_PM25',
    ascending=True
).head(10)

display(bottom_10_polluted)
```
